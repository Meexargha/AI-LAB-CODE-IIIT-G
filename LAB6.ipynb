```python
#Create a one-dimensional NumPy array containing the first 20 natural numbers. Compute its mean and standard deviation.
import numpy as np

a = np.arange(1, 21)

print("Array:", a)
print("Mean:", np.mean(a))
print("Standard Deviation:", np.std(a))
```

```python
#Create a 3 × 4 NumPy array using arange and reshape. Extract the second column using slicing.
import numpy as np

a = np.arange(12).reshape(3, 4)

print("Array:")
print(a)

print("Second column:")
print(a[:, 1])
```

```python
#Given a 2 × 3 array, repeat each row twice and each column three times.
import numpy as np

a = np.array([[1, 2, 3],
              [4, 5, 6]])

row_repeat = np.repeat(a, 2, axis=0)
col_repeat = np.repeat(a, 3, axis=1)

print("Rows repeated twice:")
print(row_repeat)

print("Columns repeated three times:")
print(col_repeat)
```

```python
#Create two NumPy arrays of shapes (3,1) and (1,4). Add them using broadcasting and report the resulting shape.
import numpy as np

a = np.array([[1],
              [2],
              [3]])

b = np.array([[10, 20, 30, 40]])

c = a + b

print("Result:")
print(c)

print("Resulting shape:", c.shape)
```
