```python
import numpy as np
import matplotlib.pyplot as plt
import keras

from keras.datasets import reuters
from keras import models
from keras import layers

# Assignemnet-1
# =========================================================
# 1. LOAD REUTERS DATASET
# =========================================================

(train_data, train_labels), (test_data, test_labels) = reuters.load_data(
    num_words=10000
)

print("Training samples:", len(train_data))
print("Testing samples:", len(test_data))
print("Number of classes:", len(np.unique(train_labels)))


# =========================================================
# 2. VECTORIZE INPUT DATA
# =========================================================

def multi_hot_encode(sequences, num_classes=10000):
    results = np.zeros((len(sequences), num_classes))

    for i, sequence in enumerate(sequences):
        results[i, sequence] = 1.0

    return results


x_train = multi_hot_encode(train_data, num_classes=10000)
x_test = multi_hot_encode(test_data, num_classes=10000)

print("x_train shape:", x_train.shape)
print("x_test shape:", x_test.shape)


# =========================================================
# 3. KEEP LABELS AS INTEGER LABELS
# =========================================================

y_train = train_labels
y_test = test_labels

print("First 10 labels:")
print(y_train[:10])


# =========================================================
# 4. BUILD MODEL
# =========================================================

model = models.Sequential([
    layers.Dense(64, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(46, activation="softmax")
])

model.build(input_shape=(None, 10000))

model.summary()


# =========================================================
# 5. TOP-3 ACCURACY
# =========================================================

top_3_accuracy = keras.metrics.SparseTopKCategoricalAccuracy(
    k=3,
    name="top_3_accuracy"
)


# =========================================================
# 6. COMPILE MODEL
# =========================================================

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=[
        "accuracy",
        top_3_accuracy
    ]
)


# =========================================================
# 7. VALIDATION SPLIT
# =========================================================

x_val = x_train[:1000]
partial_x_train = x_train[1000:]

y_val = y_train[:1000]
partial_y_train = y_train[1000:]


# =========================================================
# 8. TRAIN MODEL FOR 20 EPOCHS
# =========================================================

history = model.fit(
    partial_x_train,
    partial_y_train,
    epochs=20,
    batch_size=512,
    validation_data=(x_val, y_val)
)


# =========================================================
# 9. PLOT LOSS
# =========================================================

loss = history.history["loss"]
val_loss = history.history["val_loss"]

epochs = range(1, len(loss) + 1)

plt.figure(figsize=(8, 5))

plt.plot(
    epochs,
    loss,
    "r--",
    label="Training loss"
)

plt.plot(
    epochs,
    val_loss,
    "b-",
    label="Validation loss"
)

plt.title("Training and validation loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.xticks(epochs)
plt.legend()

plt.show()


# =========================================================
# 10. PLOT ACCURACY
# =========================================================

accuracy = history.history["accuracy"]
val_accuracy = history.history["val_accuracy"]

plt.figure(figsize=(8, 5))

plt.plot(
    epochs,
    accuracy,
    "r--",
    label="Training accuracy"
)

plt.plot(
    epochs,
    val_accuracy,
    "b-",
    label="Validation accuracy"
)

plt.title("Training and validation accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.xticks(epochs)
plt.legend()

plt.show()


# =========================================================
# 11. PLOT TOP-3 ACCURACY
# =========================================================

top3 = history.history["top_3_accuracy"]
val_top3 = history.history["val_top_3_accuracy"]

plt.figure(figsize=(8, 5))

plt.plot(
    epochs,
    top3,
    "r--",
    label="Training top-3 accuracy"
)

plt.plot(
    epochs,
    val_top3,
    "b-",
    label="Validation top-3 accuracy"
)

plt.title("Training and validation top-3 accuracy")
plt.xlabel("Epochs")
plt.ylabel("Top-3 accuracy")
plt.xticks(epochs)
plt.legend()

plt.show()


# =========================================================
# 12. RETRAIN MODEL FROM SCRATCH
# =========================================================

model = models.Sequential([
    layers.Dense(64, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(46, activation="softmax")
])

model.build(input_shape=(None, 10000))


# =========================================================
# 13. COMPILE NEW MODEL
# =========================================================

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# =========================================================
# 14. TRAIN FOR 9 EPOCHS
# =========================================================

model.fit(
    x_train,
    y_train,
    epochs=9,
    batch_size=512
)


# =========================================================
# 15. TEST THE MODEL
# =========================================================

results = model.evaluate(
    x_test,
    y_test
)

print("\nTest Loss:", results[0])
print("Test Accuracy:", results[1])


# =========================================================
# 16. GENERATE PREDICTIONS
# =========================================================

predictions = model.predict(x_test)

print("\nPrediction shape:", predictions.shape)


# =========================================================
# 17. CHECK FIRST PREDICTION
# =========================================================

print("\nFirst prediction:")
print(predictions[0])

print("\nSum of probabilities:")
print(np.sum(predictions[0]))

print("\nPredicted class:")
print(np.argmax(predictions[0]))

print("\nActual class:")
print(y_test[0])


# =========================================================
# 18. FIRST 10 PREDICTIONS
# =========================================================

predicted_classes = np.argmax(
    predictions,
    axis=1
)

print("\nPredicted classes:")
print(predicted_classes[:10])

print("\nActual classes:")
print(y_test[:10])
```

```python
#assignemnet-2
import numpy as np
import matplotlib.pyplot as plt
import keras

from keras.datasets import reuters
from keras import models
from keras import layers


# =========================================================
# 1. LOAD REUTERS DATASET
# =========================================================

(train_data, train_labels), (test_data, test_labels) = reuters.load_data(
    num_words=10000
)

print("Training samples:", len(train_data))
print("Testing samples:", len(test_data))
print("Number of classes:", len(np.unique(train_labels)))


# =========================================================
# 2. VECTORIZE INPUT DATA
# =========================================================

def multi_hot_encode(sequences, num_classes=10000):
    results = np.zeros((len(sequences), num_classes))

    for i, sequence in enumerate(sequences):
        results[i, sequence] = 1.0

    return results


x_train = multi_hot_encode(train_data, 10000)
x_test = multi_hot_encode(test_data, 10000)


# =========================================================
# 3. KEEP LABELS AS INTEGER LABELS
# =========================================================

y_train = train_labels
y_test = test_labels


# =========================================================
# 4. CREATE VALIDATION DATA
# =========================================================

x_val = x_train[:1000]
partial_x_train = x_train[1000:]

y_val = y_train[:1000]
partial_y_train = y_train[1000:]


# =========================================================
# =========================================================
# A) INFORMATION BOTTLENECK = 4 UNITS
# =========================================================
# =========================================================

print("\n===================================")
print("MODEL A: 4-DIMENSIONAL BOTTLENECK")
print("===================================")


model_4 = models.Sequential([
    layers.Dense(4, activation="relu"),
    layers.Dense(4, activation="relu"),
    layers.Dense(46, activation="softmax")
])

model_4.build(input_shape=(None, 10000))


# Compile using sparse categorical crossentropy
model_4.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# Train
history_4 = model_4.fit(
    partial_x_train,
    partial_y_train,
    epochs=20,
    batch_size=512,
    validation_data=(x_val, y_val)
)


# Get best validation accuracy
best_val_accuracy_4 = max(
    history_4.history["val_accuracy"]
)

print("\nBest Validation Accuracy (4 units):",
      best_val_accuracy_4)


# =========================================================
# =========================================================
# B) INFORMATION BOTTLENECK = 20 UNITS
# =========================================================
# =========================================================

print("\n===================================")
print("MODEL B: 20-DIMENSIONAL BOTTLENECK")
print("===================================")


model_20 = models.Sequential([
    layers.Dense(20, activation="relu"),
    layers.Dense(20, activation="relu"),
    layers.Dense(46, activation="softmax")
])

model_20.build(input_shape=(None, 10000))


# Compile
model_20.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# Train
history_20 = model_20.fit(
    partial_x_train,
    partial_y_train,
    epochs=20,
    batch_size=512,
    validation_data=(x_val, y_val)
)


# Get best validation accuracy
best_val_accuracy_20 = max(
    history_20.history["val_accuracy"]
)

print("\nBest Validation Accuracy (20 units):",
      best_val_accuracy_20)


# =========================================================
# 5. COMPARE VALIDATION ACCURACIES
# =========================================================

print("\n===================================")
print("FINAL COMPARISON")
print("===================================")

print(
    "Best Validation Accuracy - 4 units:",
    best_val_accuracy_4
)

print(
    "Best Validation Accuracy - 20 units:",
    best_val_accuracy_20
)


# =========================================================
# 6. PLOT VALIDATION ACCURACY
# =========================================================

epochs_4 = range(1, len(history_4.history["val_accuracy"]) + 1)
epochs_20 = range(1, len(history_20.history["val_accuracy"]) + 1)

plt.figure(figsize=(8, 5))

plt.plot(
    epochs_4,
    history_4.history["val_accuracy"],
    label="4-dimensional bottleneck"
)

plt.plot(
    epochs_20,
    history_20.history["val_accuracy"],
    label="20-dimensional bottleneck"
)

plt.title("Validation Accuracy: 4 vs 20 Units")
plt.xlabel("Epochs")
plt.ylabel("Validation Accuracy")
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from keras.datasets import reuters
from keras import models
from keras import layers


# =========================================================
# 1. LOAD REUTERS DATASET
# =========================================================

(train_data, train_labels), (test_data, test_labels) = reuters.load_data(
    num_words=10000
)

print("Training samples:", len(train_data))
print("Testing samples:", len(test_data))
print("Number of classes:", len(np.unique(train_labels)))


# =========================================================
# 2. VECTORIZE INPUT DATA
# =========================================================

def multi_hot_encode(sequences, num_classes=10000):
    results = np.zeros((len(sequences), num_classes))

    for i, sequence in enumerate(sequences):
        results[i, sequence] = 1.0

    return results


x_train = multi_hot_encode(train_data, 10000)
x_test = multi_hot_encode(test_data, 10000)


# =========================================================
# 3. KEEP LABELS AS INTEGER LABELS
# =========================================================

y_train = train_labels
y_test = test_labels


# =========================================================
# 4. CREATE VALIDATION DATA
# =========================================================

x_val = x_train[:1000]
partial_x_train = x_train[1000:]

y_val = y_train[:1000]
partial_y_train = y_train[1000:]


# =========================================================
# A) SINGLE INTERMEDIATE LAYER
# =========================================================

print("\n====================================")
print("MODEL A: SINGLE INTERMEDIATE LAYER")
print("====================================")

model_1 = models.Sequential([
    layers.Dense(64, activation="relu"),
    layers.Dense(46, activation="softmax")
])

model_1.build(input_shape=(None, 10000))

model_1.summary()


# Compile
model_1.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# Train
history_1 = model_1.fit(
    partial_x_train,
    partial_y_train,
    epochs=20,
    batch_size=512,
    validation_data=(x_val, y_val)
)


# Best validation accuracy
best_val_accuracy_1 = max(
    history_1.history["val_accuracy"]
)

print(
    "\nBest Validation Accuracy - Single Layer:",
    best_val_accuracy_1
)


# =========================================================
# B) THREE INTERMEDIATE LAYERS
# =========================================================

print("\n====================================")
print("MODEL B: THREE INTERMEDIATE LAYERS")
print("====================================")

model_3 = models.Sequential([
    layers.Dense(64, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(46, activation="softmax")
])

model_3.build(input_shape=(None, 10000))

model_3.summary()


# Compile
model_3.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# Train
history_3 = model_3.fit(
    partial_x_train,
    partial_y_train,
    epochs=20,
    batch_size=512,
    validation_data=(x_val, y_val)
)


# Best validation accuracy
best_val_accuracy_3 = max(
    history_3.history["val_accuracy"]
)

print(
    "\nBest Validation Accuracy - Three Layers:",
    best_val_accuracy_3
)


# =========================================================
# 5. COMPARE VALIDATION ACCURACIES
# =========================================================

print("\n====================================")
print("FINAL COMPARISON")
print("====================================")

print(
    "Single intermediate layer:",
    best_val_accuracy_1
)

print(
    "Three intermediate layers:",
    best_val_accuracy_3
)


# =========================================================
# 6. PLOT VALIDATION ACCURACY
# =========================================================

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 21),
    history_1.history["val_accuracy"],
    label="1 intermediate layer"
)

plt.plot(
    range(1, 21),
    history_3.history["val_accuracy"],
    label="3 intermediate layers"
)

plt.title("Validation Accuracy: 1 vs 3 Intermediate Layers")
plt.xlabel("Epochs")
plt.ylabel("Validation Accuracy")
plt.legend()

plt.show()
```
